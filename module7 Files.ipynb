```python
# 21.Write a Python program that reads a file and prints the number of occurrences of a specific word.
word = input("Enter the word to count: ").lower() #take input in lower case, since Python is case sensitive

with open("test.txt", "r") as file:
    content = file.read().lower()  # read entire file and convert to lowercase

new_count = content.count(word)   # Count the word

# Print the result
print(f"The word '{word}' occurred {new_count} times in the file.")
```
